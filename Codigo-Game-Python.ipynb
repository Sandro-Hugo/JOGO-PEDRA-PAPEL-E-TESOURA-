 ## Script Python

Agora, vamos traduzir o pseudocódigo para um script Python.

```python
# Jogo Pedra, Papel e Tesoura

print("--- Jogo Pedra, Papel e Tesoura ---")
print("Escolha: pedra, papel ou tesoura")

validos = ("pedra", "papel", "tesoura")

# Jogadores fazem suas escolhas
jogador1 = input("Jogador 1: ")
jogador2 = input("Jogador 2: ")

# Deixar as escolhas em letras minúsculas
jogador1 = jogador1.lower().strip()
jogador2 = jogador2.lower().strip()

# Mostrar as escolhas
print("Jogador 1 escolheu:", jogador1)
print("Jogador 2 escolheu:", jogador2)

# Verificando se o input é válido
if jogador1 not in validos or jogador2 not in validos:
    print(f"Por gentileza, escolha as opções válidas: {validos}")

# Verificar o resultado
elif jogador1 == jogador2:
    print("Resultado: Empate!")

elif jogador1 == "pedra" and jogador2 == "tesoura":
    print("Resultado: Jogador 1 venceu!")

elif jogador1 == "tesoura" and jogador2 == "papel":
    print("Resultado: Jogador 1 venceu!")

elif jogador1 == "papel" and jogador2 == "pedra":
    print("Resultado: Jogador 1 venceu!")

else:
    print("Resultado: Jogador 2 venceu!")

print("--- Fim de Jogo ---")
```
