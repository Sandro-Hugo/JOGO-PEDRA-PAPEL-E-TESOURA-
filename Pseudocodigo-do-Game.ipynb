**CRIANDO O PSEUDOCÓDIGO DO JOGO**

# JOGO PEDRA, PAPEL E TESOURA

INÍCIO

// 1. Apresentação

MOSTRAR "--- Jogo Pedra, Papel e Tesoura ---"

MOSTRAR "Escolha: pedra, papel ou tesoura"

// 2. Jogadores fazem suas escolhas

LER escolha do Jogador 1

LER escolha do Jogador 2

// 3. Ajustar as escolhas

CONVERTER escolha do Jogador 1 para letras minúsculas

REMOVER espaços extras da escolha do Jogador 1

CONVERTER escolha do Jogador 2 para letras minúsculas

REMOVER espaços extras da escolha do Jogador 2

// 4. Mostrar as escolhas

MOSTRAR "Jogador 1 escolheu:", escolha do Jogador 1

MOSTRAR "Jogador 2 escolheu:", escolha do Jogador 2

// 5. Verificar o resultado

SE Jogador 1 escolher a mesma coisa que Jogador 2
MOSTRAR "Resultado: Empate!"

SENÃO SE Jogador 1 escolher "pedra"
E Jogador 2 escolher "tesoura"
MOSTRAR "Resultado: Jogador 1 venceu!"

SENÃO SE Jogador 1 escolher "tesoura"
E Jogador 2 escolher "papel"
MOSTRAR "Resultado: Jogador 1 venceu!"

SENÃO SE Jogador 1 escolher "papel"
E Jogador 2 escolher "pedra"
MOSTRAR "Resultado: Jogador 1 venceu!"

SENÃO
MOSTRAR "Resultado: Jogador 2 venceu!"

// 6. Encerramento

MOSTRAR "--- Fim de Jogo ---"

FIM
